```python
import sys
sys.path.insert(0, '..')
from notebooks.nb_support import services, sample_document, arun, ROOT
ct, MODE = services()
print('mode:', MODE)
```

```python
OWNER = 'notebook_demo'
name, data = sample_document()
arun(ct.store.delete_namespace(OWNER))
doc = arun(ct.ingestion.upload(OWNER, name, data, None))
print(doc)
```

```python
print('vector ids:', [f"{doc['doc_id'][:8]}…:{i}" for i in range(doc['num_chunks'])][:5])
print('queryable:', arun(ct.store.exists(OWNER, f"{doc['doc_id']}:{doc['num_chunks'] - 1}")))
print('registry:', ct.db.list_documents(OWNER)[0]['num_chunks'], 'chunks')
```
