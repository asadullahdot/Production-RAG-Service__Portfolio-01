```python
import sys
sys.path.insert(0, '..')
from notebooks.nb_support import services, sample_document, arun, reset_owner, ROOT
ct, MODE = services()
print('mode:', MODE)
```

```python
from app.retrieval import retrieve
OWNER = 'notebook_demo'
name, data = sample_document()
reset_owner(ct, OWNER)
for d in ct.db.list_documents(OWNER):
    ct.db.delete_document(OWNER, d['doc_id'])
arun(ct.ingestion.upload(OWNER, name, data, None))
res = arun(retrieve(OWNER, 'What is the refund policy for returns?', embedder=ct.openai, store=ct.store, db=ct.db, settings=ct.settings, threshold=0.0))
for c in res.raw:
    print(f'[{c.n}] score={c.score:.3f} page={c.page} {c.text[:90]!r}')
```
