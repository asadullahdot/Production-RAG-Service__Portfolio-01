```python
from app.ingestion.extraction import extract
from app.ingestion.chunking import chunk_sections
name, data = sample_document()
ex = extract(data, name.rsplit('.', 1)[-1])
for label, (size, overlap) in {'A': (1000, 200), 'B': (500, 100)}.items():
    chunks = chunk_sections(ex.sections, size, overlap)
    print(label, size, overlap, '-> chunks:', len(chunks), 'max len:', max(len(c.text) for c in chunks),
          'pages:', sorted({c.page for c in chunks}))
```
